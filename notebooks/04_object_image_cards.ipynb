```python
#generating custom cards to view objects and their related historical and astrological data
##install Pillow for image handling
#code adapted from https://www.reddit.com/r/Python/comments/1jawsk6/createintrocards_convert_a_dataset_of_peoples/

from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4, landscape
from reportlab.lib.units import inch
from reportlab.lib import colors
from google.colab import files
import os
import shutil

#configuring the cards
def create_intro_card(row, filename):
  c = canvas.Canvas(filename, pagesize=(4 * inch, 2.5 * inch))
  width, height = 4 * inch, 2.5 * inch

  #background
  c.setFillColor(colors.whitesmoke)
  c.rect(0, 0, width, height, fill=1, stroke=0)

  #Title
  c.setFillColor(colors.darkred)
  c.setFont("Helvetica-Bold", 18)
  c.drawString(0.2 * inch, 1.8 * inch, row['Title'])

 # Rest of Content Font and Fill
  c.setFillColor(colors.black)
  c.setFont("Helvetica", 10)

  #Rest of Content
  c.drawString(0.2 * inch, 1.7 * inch, f"Artist: {row['Artist.Display.Name']}")
  c.drawString(0.2 * inch, 1.45 * inch, f"Date: {row['Object.Begin.Date']}")
  c.drawString(0.2 * inch, 1.2 * inch, f"Country: {row['Country']}")


  c.setFont("Helvetica-Oblique", 9)
  c.drawString(0.2 * inch, 0.95 * inch, f"Incident: {row['Historical Incident']}")
  c.drawString(0.2 * inch, 0.7 * inch, f"Impact: {row['Impact']}")

  c.save()

#Generate cards for each database entry
os.makedirs('my_cards', exist_ok=True)

for index, row in df.iterrows():
    clean_title = "".join(x for x in str(row['Object.ID']) if x.isalnum())
    filename = f"Card_{clean_title}.pdf"

    create_intro_card(row, filename)
    print(f"Generated: {filename}")

#Downloading files
# Zip the folder and download once
shutil.make_archive('all_intro_cards', 'zip', 'my_cards')
files.download('all_intro_cards.zip')
```
